```python
import tensorflow as tf
import numpy as np
import json
import os
from tensorflow.keras.preprocessing import image

MODEL_PATH = r"D:\CODE\MiniProject\ml-service\models\leaf_model.h5"
CLASS_PATH = r"D:\CODE\MiniProject\ml-service\models\class_names.json"
IMG_PATH   = r"D:\CODE\MiniProject\ml-service\data\leaf_images\small_dataset\test_leaf.jpg"

IMG_SIZE = 224

# Safety checks
assert os.path.exists(MODEL_PATH), "❌ Model file missing"
assert os.path.exists(CLASS_PATH), "❌ class_names.json missing"
assert os.path.exists(IMG_PATH), "❌ Image file missing"

# Load model
model = tf.keras.models.load_model(MODEL_PATH)

# Load class names
with open(CLASS_PATH, "r") as f:
    class_names = json.load(f)

# Load & preprocess image
img = image.load_img(IMG_PATH, target_size=(IMG_SIZE, IMG_SIZE))
img_array = image.img_to_array(img)
img_array = np.expand_dims(img_array, axis=0) / 255.0

# Predict
pred = model.predict(img_array)
class_index = np.argmax(pred)
confidence = np.max(pred)

print("🌿 Predicted Disease :", class_names[class_index])
print("📊 Confidence       :", round(confidence * 100, 2), "%")
```
